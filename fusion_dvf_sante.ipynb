```python
import pandas as pd

# Charger les deux jeux de données
dvf = pd.read_csv("dvf_filtre.csv")
sante = pd.read_csv("sante_score.csv")

# 🔧 Harmoniser les types
dvf["code_postal"] = dvf["code_postal"].astype(str).str.split('.').str[0].str.zfill(5)
sante["code_commune"] = sante["code_commune"].astype(str).str.zfill(5)

# Vérification
print(dvf["code_postal"].head())
print(sante["code_commune"].head())

# Fusion
df_final = pd.merge(
    dvf,
    sante,
    how="left",
    left_on="code_postal",
    right_on="code_commune"
)

print("✅ Fusion réalisée :", df_final.shape)
display(df_final.head(10))

# Sauvegarde
df_final.to_csv("dvf_sante_fusion.csv", index=False)
print("💾 Fichier fusionné enregistré sous : dvf_sante_fusion.csv")
```
