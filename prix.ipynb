```python
# === 1️⃣ Import des bibliothèques ===
import pandas as pd
import ipywidgets as widgets
from ipywidgets import interact
import matplotlib.pyplot as plt

# === 2️⃣ Chargement du fichier nettoyé ===
df = pd.read_csv("dvf_clean.csv")

# Conversion utile pour éviter les erreurs
df["valeur_fonciere"] = pd.to_numeric(df["valeur_fonciere"], errors="coerce")
df["nombre_pieces_principales"] = pd.to_numeric(df["nombre_pieces_principales"], errors="coerce")

# === 3️⃣ Widgets interactifs ===
budget_slider = widgets.IntSlider(
    value=350000, min=50000, max=600000, step=10000,
    description='💰 Budget max', style={'description_width': 'initial'},
    continuous_update=False
)

pieces_slider = widgets.IntSlider(
    value=3, min=1, max=8, step=1,
    description='🛏️ Pièces min', style={'description_width': 'initial'},
    continuous_update=False
)

type_dropdown = widgets.Dropdown(
    options=['Tous', 'Maison', 'Appartement'],
    value='Tous',
    description='🏠 Type de bien',
    style={'description_width': 'initial'}
)

# === 4️⃣ Fonction d’affichage ===
def filtrer_dvf(budget_max, nb_pieces, type_bien):
    # Filtrage
    df_filtre = df[
        (df["valeur_fonciere"] <= budget_max) &
        (df["nombre_pieces_principales"] >= nb_pieces)
    ].copy()

    if type_bien != "Tous":
        df_filtre = df_filtre[df_filtre["type_local"] == type_bien]

    # Affichage texte
    print(f"📊 {len(df_filtre):,} biens trouvés correspondant à vos critères")

    # Sauvegarde automatique du résultat pour les collègues
    df_filtre.to_csv("dvf_filtre.csv", index=False)
    print("✅ Données filtrées sauvegardées sous dvf_filtre.csv")

    display(df_filtre.head(10))

    # Afficher un petit aperçu du tableau
    display(df_filtre[["nom_commune", "type_local", "nombre_pieces_principales",
                       "surface_reelle_bati", "valeur_fonciere", "prix_m2"]].head(10))

    # === Graphique simple (distribution du prix au m²) ===
    if not df_filtre.empty:
        plt.figure(figsize=(7,4))
        plt.hist(df_filtre["prix_m2"], bins=30, color='skyblue', edgecolor='black')
        plt.title("Distribution du prix au m² des biens filtrés")
        plt.xlabel("Prix au m² (€)")
        plt.ylabel("Nombre de biens")
        plt.grid(alpha=0.3)
        plt.show()
    else:
        print("⚠️ Aucun bien ne correspond à ces critères.")

# === 5️⃣ Lier les widgets à la fonction ===
interact(
    filtrer_dvf,
    budget_max=budget_slider,
    nb_pieces=pieces_slider,
    type_bien=type_dropdown
);
```
